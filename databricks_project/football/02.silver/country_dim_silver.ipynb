```python
from pyspark.sql.functions import when, col
from datetime import date
from datetime import datetime
from pyspark.sql.functions import current_timestamp

df = spark.read.table("main.db_project_bronze.leauges_bronze")
df = (
    spark.read.table("main.db_project_bronze.leauges_bronze")
      .withColumn(
          "country_code",
          when(col("country_code").isNull(), "N/A")
          .otherwise(col("country_code"))
      )
      .withColumn("insert_timestamp", current_timestamp())
)
df = df.select("country", "country_code", "insert_timestamp").dropDuplicates()

df.write.mode("overwrite").saveAsTable("main.db_project_silver.leauges_silver")
```
